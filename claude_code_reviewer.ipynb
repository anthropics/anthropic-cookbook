# AI-Powered Code Reviewer with Claude

This notebook demonstrates how to use Claude as an intelligent code reviewer that analyzes Python code for:
- 🐛 **Bugs** — logic errors and potential crashes
- 🔒 **Security Issues** — vulnerabilities and bad practices
- ⚡ **Performance** — inefficiencies and optimizations
- 📖 **Readability** — code clarity and best practices

````python
import anthropic
import os
from dotenv import load_dotenv

load_dotenv()

# Set your API key in .env file as: ANTHROPIC_API_KEY=your_key_here
client = anthropic.Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))

print("✅ Claude client initialized successfully!")
````

## Core Function — Claude Code Reviewer

````python
def review_code(code: str, language: str = "Python") -> str:
    """
    Review code using Claude and return detailed feedback.
    
    Args:
        code: The source code to review
        language: Programming language (default: Python)
    
    Returns:
        Detailed code review as a string
    """
    prompt = f"""You are an expert {language} code reviewer. Review the following code and provide feedback in these sections:

1. 🐛 **Bugs & Errors** — Any logic errors or potential crashes
2. 🔒 **Security Issues** — Vulnerabilities or unsafe practices  
3. ⚡ **Performance** — Inefficiencies and how to fix them
4. 📖 **Readability** — Code clarity and best practices
5. ✅ **Improved Code** — Provide a corrected version

Code to review:
```{language.lower()}
{code}
```

Be specific, helpful, and beginner-friendly in your review."""

    message = client.messages.create(
        model="claude-haiku-4-5",
        max_tokens=1500,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    
    return message.content[0].text
````

## Example 1 — Review Buggy Code

````python
buggy_code = """
def calculate_average(numbers):
    total = 0
    for num in numbers:
        total = total + num
    average = total / len(numbers)
    return average

# Test
print(calculate_average([]))
print(calculate_average([1, 2, 3, 4, 5]))
"""

print("🔍 Reviewing code...\n")
review = review_code(buggy_code)
print(review)
````

## Example 2 — Review Code with Security Issues

````python
insecure_code = """
import sqlite3

def get_user(username):
    conn = sqlite3.connect('users.db')
    cursor = conn.cursor()
    query = "SELECT * FROM users WHERE username = '" + username + "'"
    cursor.execute(query)
    return cursor.fetchone()

user = get_user("admin' OR '1'='1")
print(user)
"""

print("🔍 Reviewing code for security issues...\n")
review = review_code(insecure_code)
print(review)
````

## Example 3 — Review Your Own Code

Paste any code you want reviewed!

````python
your_code = """
# Paste your code here!
def my_function():
    pass
"""

print("🔍 Reviewing your code...\n")
review = review_code(your_code)
print(review)
````

## Bonus — Quick Review Function

Get a one-line summary of code quality.

````python
def quick_review(code: str) -> str:
    """Get a quick one-line code quality score."""
    message = client.messages.create(
        model="claude-haiku-4-5",
        max_tokens=100,
        messages=[{
            "role": "user",
            "content": f"Rate this code quality from 1-10 and give ONE sentence reason:\n\n{code}"
        }]
    )
    return message.content[0].text

sample = """
def add(a, b):
    return a + b
"""

print("⚡ Quick Review:")
print(quick_review(sample))
````
